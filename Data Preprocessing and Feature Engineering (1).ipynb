# Fill Missing Values

```python
import pandas as pd
```

```python
# Load Titanic dataset
# loaded the Titanic dataset into a DataFrame (df)
df = pd.read_csv('titanic.csv')
```

```python
df.head()
```

```python
# Fill missing values in 'Age' column with the median
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Age'].head()
```

```python
# Fill missing values in categorical columns with the mode
categorical_columns = df.select_dtypes(include=['object']).columns
```

```python
categorical_columns
```

```python
for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
 df[col].head()
```

```python
# Check if there are any missing values left
df.isnull().sum()
```

# One-Hot Encoding and Label Encoding

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
# Load Titanic dataset
# loaded the Titanic dataset into a DataFrame (df)
df = pd.read_csv('titanic.csv')
```

```python
# Displaying the first few rows to understand the dataset
df.head()
```

```python
# Identifying categorical variables
categorical_columns = ['Sex', 'Embarked', 'Pclass']
```

```python
categorical_columns 
```

```python
# One-Hot Encoding for 'Sex' and 'Embarked'
titanic_df_encoded = pd.get_dummies(df, columns=['Sex', 'Embarked'], drop_first=True)
```

```python
titanic_df_encoded.head()
```

```python
# Label Encoding for 'Pclass'
label_encoder = LabelEncoder()
titanic_df_encoded['Pclass'] = label_encoder.fit_transform(titanic_df_encoded['Pclass'])
```

```python
titanic_df_encoded['Pclass'].head()
```

```python
# Display the resulting DataFrame
(titanic_df_encoded.head())
```

# Create New Feautures

```python
# Assuming you have loaded the Titanic dataset into a pandas DataFrame called df
df = pd.read_csv('titanic.csv')  
```

```python
df.head()
```

```python
# Feature 1: Family Size (SibSp + Parch + 1)
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
```

```python
df['FamilySize'].head()
```

```python
# Feature 2: Is Alone (1 if FamilySize == 1, else 0)
df['IsAlone'] = df['FamilySize'].apply(lambda x: 1 if x == 1 else 0)
```

```python
df['IsAlone'].head()
```

```python
# Feature 3: Age Group
df['AgeGroup'] = pd.cut(df['Age'], bins=[0, 12, 19, 60, float('inf')], labels=['Child', 'Teenager', 'Adult', 'Senior'])
```

```python
df['AgeGroup'].head()
```

```python
# Feature 4: Fare per Person (Fare divided by FamilySize)
df['FarePerPerson'] = df['Fare'] / df['FamilySize']
```

```python
df['FarePerPerson'].head()
```

```python
# Feature 5: Embarked Mode (One-hot encoding for Embarked)
df['Embarked_C'] = (df['Embarked'] == 'C').astype(int)
df['Embarked_Q'] = (df['Embarked'] == 'Q').astype(int)
df['Embarked_S'] = (df['Embarked'] == 'S').astype(int)
```

```python
df['Embarked_C'].head()
```

```python
df['Embarked_Q'].head()
```

```python
df['Embarked_S'].head()
```

```python
# Feature 6: Interaction between Age and Pclass
df['Age_Pclass_Interaction'] = df['Age'] * df['Pclass']
```

```python
df['Age_Pclass_Interaction'].head()
```

```python
# Feature 07: Fare Binned (Categorizing Fare into Low, Medium, High)
df['FareBin'] = pd.cut(df['Fare'], bins=[0, 10, 50, float('inf')], labels=['Low', 'Medium', 'High'])
```

```python
df['FareBin'].head()
```

```python
# Feature 8: Interaction between Pclass and Sex
df['Pclass_Sex'] = df['Pclass'] * (df['Sex'] == 'male').astype(int)
```

```python
df['Pclass_Sex'].head()
```

# Extract Title From The Passenger Name

```python
import pandas as pd
import re

# Sample Titanic dataset
df = pd.read_csv('titanic.csv')

# Function to extract title
def extract_title(name):
    title = re.search(r'([A-Za-z]+)\.', name)  # Search for any word followed by a period
    return title.group(1) if title else None  # If found, return the title

# Apply the function to the 'Name' column to extract the title
df['Title'] = df['Name'].apply(extract_title)

# Display the updated dataframe
print(df[['Name', 'Title']].head())
```
